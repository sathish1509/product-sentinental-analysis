# 01. Data Loading & Initial Inspection

This notebook demonstrates the initial ingestion, structural analysis, missing data inspection, and dataset sanity checks for the Product Review Sentiment Analysis pipeline.

### Objective:
1. Load raw review data from CSV and JSON formats.
2. Inspect schema, dataset shape, and sample records.
3. Check missing values and duplicate entries.
4. Produce basic statistical summaries for ratings and categories.


```python
import pandas as pd
import json
import os

print('[+] Pandas version:', pd.__version__)
```

## Step 1: Load Raw CSV Dataset


```python
csv_path = '../raw/product_reviews.csv'
df_csv = pd.read_csv(csv_path)
print(f'CSV Dataset Shape: {df_csv.shape}')
df_csv.head(5)
```

## Step 2: Load Raw JSON Fallback Dataset


```python
json_path = '../raw/sample_reviews.json'
with open(json_path, 'r') as f:
    json_data = json.load(f)

df_json = pd.DataFrame(json_data)
print(f'JSON Dataset Shape: {df_json.shape}')
df_json.head(5)
```

## Step 3: Column Names and Data Types


```python
print('--- CSV Columns & Dtypes ---')
print(df_csv.dtypes)
print('\n--- JSON Columns & Dtypes ---')
print(df_json.dtypes)
```

## Step 4: Missing Value Analysis


```python
print('--- CSV Missing Values ---')
print(df_csv.isnull().sum())
print('\n--- JSON Missing Values ---')
print(df_json.isnull().sum())
```

## Step 5: Duplicate Inspection & Rating Distribution


```python
duplicates_csv = df_csv.duplicated(subset=['product_id', 'review_text']).sum()
print(f'Duplicate review count in CSV: {duplicates_csv}')

print('\nCSV Rating Summary:')
print(df_csv['rating'].describe())
```

### Key Takeaways:
- The raw dataset contains all required fields (`product_id`, `product_name`, `category`, `review_title`, `review_text`, `rating`, `review_date`, `verified_purchase`).
- Data types are appropriate (strings, integers, dates, booleans).
- No critical missing values were found in key text or label columns.
